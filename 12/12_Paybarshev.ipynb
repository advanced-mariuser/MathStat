**1. [# 10]** Предполагается, что полиграф, используемый в уголовных расследованиях, позволяет определить, лжет человек или говорит правду. Однако, как видно из следующего примера, эта процедура не является безошибочной. Опытному полиграфологу было предложено вынести общее заключение по каждой из 280 записей, из которых 140 касались подозреваемых в совершении преступления и 140 – невиновных. Результаты приведены в таблице. Мы рассматриваем каждое судебное решение как задачу проверки гипотез, где нулевая гипотеза соответствует «подозреваемый невиновен», а альтернативная гипотеза – «подозреваемый виновен». Оцените вероятности ошибок I типа и II типа, которые применимы к данному методу проверки на полиграфе, на основе таблицы.

```python
true_status = np.array([[131, 9], [15, 125]])
alpha = true_status[0,1] / (true_status[0,0] + true_status[0,1])
beta = true_status[1,0] / (true_status[1,0] + true_status[1,1])
print(f"Вероятность ошибки I типа (α): {alpha:.4f}")
print(f"Вероятность ошибки II типа (β): {beta:.4f}")
```

**2. [# 30]** Соревнования по конькобежному спорту на дистанции 1500 метров на зимних Олимпийских играх 2002 года. Количество заездов, выигранных фигуристами, стартовавшими на внешней дорожке, моделируется случайной величиной $X$ с распределением $Bin(23, p)$. Рассмотрим этот вопрос, проверяя нулевую гипотезу <br><br>
$H_0$: $p = 1/2$ против $H_1: p > 1/2$, используя $X$ в качестве тестовой статистики. Из 23
завершенных заездов 15 были выиграны фигуристами, стартовавшими на внешней
дорожке. <br><br>
a. Вычислите значение $p$, соответствующее $x = 15$, и сообщите о своем заключении,
если мы проведем тест на уровне 0.05.<br><br>
b. Определите критическую область, соответствующую уровню значимости $\alpha = 0.05$.<br><br>
c. Вычислите вероятность совершения ошибки типа I, если мы основываем наше правило принятия решений на критической области, определенной в пункте b. <br><br>
d. Используйте нормальное приближение, чтобы определить вероятность совершения
ошибки второго типа для случая $p = 0.6$, если мы основываем наше правило принятия
решения на критической области, определенной в b.

```python
n = 23
x = 15
p0 = 0.5
alpha = 0.05

# a. Вычисление p-значения и принятие решения
print("a.")
p_value = 1 - binom.cdf(x-1, n, p0)  # Односторонний тест, p-value для правого хвоста
print(f"p-значение: {p_value:.4f}")
if p_value < alpha:
    print("Отвергаем H0 в пользу H1 на уровне значимости α=0.05")
else:
    print("Не отвергаем H0 на уровне значимости α=0.05")

# b. Определение критической области
print("\nb.")
critical_value = binom.ppf(1-alpha, n, p0)
print(f"Критическая область: x > {critical_value}")

# c. Вероятность ошибки I типа
print("\nc.")
alpha_actual = 1 - binom.cdf(int(critical_value)-1, n, p0)
print(f"Вероятность ошибки I типа (α): {alpha_actual:.4f}")

# d. Вероятность ошибки II типа (приближенно, нормальное приближение)
print("\nd.")
p1 = 0.6

mean = n * p1
std = np.sqrt(n * p1 * (1 - p1))
z_critical = (critical_value - mean - 0.5) / std
beta_approx = norm.cdf(z_critical)
print(f"Вероятность ошибки II типа (β): {beta_approx:.4f}")
```

**3. [# 15]** Чтобы проверить гипотезу о том, что шансы лошади на победу в восьмикратном
забеге на кольцевой трассе зависят от ее положения в стартовом составе, было
зафиксировано стартовое положение каждого из 144 победителей. Оказалось, что 29 из
этих победителей заняли стартовую позицию номер один (ближайшую к рельсам на
внутренней дорожке). Мы моделируем количество победителей со стартовой позицией
номер один с помощью случайной величины $T$ с распределением $Bin(144, p)$.
Проверьте гипотезу $H_0: p = 1/8$ против $H_1: p > 1/8$ при уровне $\alpha = 0.01$, используя $T$
в качестве тестовой статистики.

```python
n = 144
x = 29
p0 = 1/8
alpha = 0.01

p_value = 1 - binom.cdf(x-1, n, p0)
print(f"p-значение: {p_value:.4f}")
if p_value < alpha:
    print("Отвергаем H0 в пользу H1 на уровне значимости α = 0.01")
else:
    print("Не отвергаем H0 на уровне значимости α = 0.01")
```
