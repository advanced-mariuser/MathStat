[# 10]** В ходе исследования, посвященного задержкам движения поездов метро, один из участников поинтересовался, отличаются ли задержки прибытия поездов в часы пик в большей степени, чем в обычные часы. Наблюдаемые задержки прибытия в часы пик моделируются как реализации случайной выборки из распределения с дисперсией $\sigma_1^2$, и аналогично наблюдаемые задержки прибытия в тихие часы соответствуют распределению с дисперсией $\sigma_2^2$. Проверяется нулевая гипотеза $H_0: \sigma_1 = \sigma_2$. Что бы вы выбрали в качестве альтернативной гипотезы?

$H_1$: $\sigma_1 > \sigma_2$ - это альтернативная гипотеза означает, что дисперсия задержек в часы пик больше, чем в тихие часы. Это может означать, что поезда метро подвержены более значительным задержкам в периоды пиковой нагрузки.

**2. [# 10]** В среднем за сентябрь в северном городе, рождается 1472 ребенка. В январе 1977 года снежная буря парализовала город. Девять месяцев спустя, в сентябре 1977 года, число зарегистрированных рождений составило 1718. Можно ли объяснить увеличение на 246 случаев случайностью? Чтобы исследовать это, число рождений в сентябре было смоделировано с помощью пуассоновской случайной величины с параметром $\mu$, и мы проверили $H_0: \mu = 1472$. Какую бы вы выбрали альтернативную гипотезу?

$H_1$: $\mu > 1472$ - эта альтернативная гипотеза предполагает, что среднее число рождений ($\mu$) в сентябре 1977 года больше, чем среднее число рождений в обычные месяцы (1472).

**3. [# 20]** Число рождений в сентябре моделируется пуассоновской случайной величиной $T$ с параметром $\mu$, который представляет ожидаемое число рождений. Предположим, что кто-то использует $T$ для проверки нулевой гипотезы $H_0: \mu = 1472$ и что кто-то решает отклонить $H_0$ на основе наблюдения значения $t = 1718$.


**a.** В каком направлении значения $T$ свидетельствуют против $H_0$ (и в пользу $H_1$)?

Значения $T$, отличающиеся от предполагаемого среднего $\mu = 1472$, свидетельствуют против нулевой гипотезы $H_0$ и в пользу альтернативной гипотезы $H_1$. В данном случае, поскольку наблюдаемое значение $t = 1718$ больше, чем ожидаемое среднее $\mu = 1472$, это говорит в пользу того, что среднее количество рождений в сентябре действительно больше, чем предполагалось по нулевой гипотезе.

**b.** Вычислите значение $p$, соответствующее $t = 1718$, где вы можете использовать тот факт, что распределение $T$ может быть аппроксимировано распределением $N(\mu, \mu)$.

```python
mu = 1472
t = 1718

std_dev = mu

z_score = (t - mu) / std_dev

p_value = 1 - stats.norm.cdf(z_score)

print("Значение p для t = 1718:", p_value)
```
