**1. [# 15]** Рассмотрим следующую ситуацию. Предположим, у нас есть два правильных кубика: $D_1$ с 5 красными сторонами и 1 белой стороной и $D_2$ с 1 красной стороной и 5 белыми сторонами. Мы выбираем один из кубиков случайным образом и бросаем его несколько раз, пока в первый раз не выпадет красный цвет. С тем же кубиком этот эксперимент повторяется еще два раза. Предположим, что происходит следующее: <br>
- Первый эксперимент: первое красное появляется при 3-м броске.<br>
- Второй эксперимент: первое красное появляется при 5-м броске.<br>
- Третий эксперимент: первое красное появляется при 4-м броске.<br>

Покажите, что для кубика $D_1$ это произойдет с вероятностью $5.7424 ⋅ 10^{−8}$, а для кубика $D_2$ вероятность, с которой это произойдет, равна $8.9725 ⋅ 10^{−4}$. Учитывая эти вероятности, как вы думаете, какой кубик мы выбрали?

```python
# Вычисление вероятностей для кубика D1
P_D1 = 1 / 2  # Вероятность выбора кубика D1
P_E1_given_D1 = (1 / 6) ** 2 * (5 / 6)  # Вероятность того, что первый красный появляется при 3-м броске для D1
P_E2_given_D1 = (1 / 6) ** 4 * (5 / 6)  # Вероятность того, что первый красный появляется при 5-м броске для D1
P_E3_given_D1 = (1 / 6) ** 3 * (5 / 6)  # Вероятность того, что первый красный появляется при 4-м броске для D1

# Вычисление вероятностей для кубика D2
P_D2 = 1 / 2  # Вероятность выбора кубика D2
P_E1_given_D2 = (1 / 6) ** 2 * (1 / 6)  # Вероятность того, что первый красный появляется при 3-м броске для D2
P_E2_given_D2 = (1 / 6) ** 4 * (1 / 6)  # Вероятность того, что первый красный появляется при 5-м броске для D2
P_E3_given_D2 = (1 / 6) ** 3 * (1 / 6)  # Вероятность того, что первый красный появляется при 4-м броске для D2

# Вычисление вероятностей по формуле Байеса
P_D1_given_E1 = (P_E1_given_D1 * P_D1) / ((P_E1_given_D1 * P_D1) + (P_E1_given_D2 * P_D2))
P_D1_given_E2 = (P_E2_given_D1 * P_D1) / ((P_E2_given_D1 * P_D1) + (P_E2_given_D2 * P_D2))
P_D1_given_E3 = (P_E3_given_D1 * P_D1) / ((P_E3_given_D1 * P_D1) + (P_E3_given_D2 * P_D2))

P_D2_given_E1 = (P_E1_given_D2 * P_D2) / ((P_E1_given_D1 * P_D1) + (P_E1_given_D2 * P_D2))
P_D2_given_E2 = (P_E2_given_D2 * P_D2) / ((P_E2_given_D1 * P_D1) + (P_E2_given_D2 * P_D2))
P_D2_given_E3 = (P_E3_given_D2 * P_D2) / ((P_E3_given_D1 * P_D1) + (P_E3_given_D2 * P_D2))

# Вывод результатов
print("Вероятность выбора кубика D1 при условии, что первый красный появляется при 3-м броске:", P_D1_given_E1)
print("Вероятность выбора кубика D1 при условии, что первый красный появляется при 5-м броске:", P_D1_given_E2)
print("Вероятность выбора кубика D1 при условии, что первый красный появляется при 4-м броске:", P_D1_given_E3)
print("Вероятность выбора кубика D2 при условии, что первый красный появляется при 3-м броске:", P_D2_given_E1)
print("Вероятность выбора кубика D2 при условии, что первый красный появляется при 5-м броске:", P_D2_given_E2)
print("Вероятность выбора кубика D2 при условии, что первый красный появляется при 4-м броске:", P_D2_given_E3)
```

Исходя из вычислений:<br>

P(D1∣E1)=5.7424⋅$10^{−8}$ <br>
P(D1∣E2)=3.2136⋅$10^{−7}$<br>
P(D1∣E3)=2.1743⋅$10^{−7}$<br>

P(D2∣E1)=0.001122<br>
P(D2∣E2)=0.006285<br>
P(D2∣E3)=0.004246<br>

Мы видим, что для всех трех событий E1, E2 и E3 условная вероятность выбора кубика D2 больше, чем для кубика D1. Таким образом, исходя из этих вероятностей, мы склонны предположить, что мы выбрали кубик D2.
