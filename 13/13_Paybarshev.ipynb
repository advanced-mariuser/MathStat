1. [# 10] Мы проверяем t-критерий для нулевой гипотезы H_0:μ=10 с помощью набора данных, состоящего из n=16 элементов с выборочным средним значением 11 и выборочной дисперсией 4. Мы используем уровень значимости 0.05. <br>
а. Следует ли нам отклонить нулевую гипотезу в пользу H_1:μ≠10?<br>
б. Что, если мы проведем проверку по H_1:μ>10?<br>


```python
n = 16
sample_mean = 11
sample_variance = 4
sample_std = np.sqrt(sample_variance)
mu0 = 10
alpha = 0.05

# Вычисление t-статистики
t_stat = (sample_mean - mu0) / (sample_std / np.sqrt(n))

# Количество степеней свободы
df = n - 1

# a. Проверка гипотезы H1: μ ≠ 10 (двусторонний тест)
t_critical_2tailed = stats.t.ppf(1 - alpha/2, df)
p_value_2tailed = (1 - stats.t.cdf(abs(t_stat), df)) * 2

print("a.")
print(f"t-статистика: {t_stat:.4f}")
print(f"Критическое значение : ±{t_critical_2tailed:.4f}")
print(f"p-значение : {p_value_2tailed:.4f}")
if abs(t_stat) > t_critical_2tailed:
    print("Отклоняем нулевую гипотезу H0 в пользу H1: μ ≠ 10 на уровне значимости 0.05")
else:
    print("Не можем отклонить нулевую гипотезу H0 на уровне значимости 0.05")

# b. Проверка гипотезы H1: μ > 10 (правосторонний тест)
t_critical_1tailed = stats.t.ppf(1 - alpha, df)
p_value_1tailed = 1 - stats.t.cdf(t_stat, df)

print("\nb.")
print(f"t-статистика: {t_stat:.4f}")
print(f"Критическое значение : {t_critical_1tailed:.4f}")
print(f"p-значение : {p_value_1tailed:.4f}")
if t_stat > t_critical_1tailed:
    print("Отклоняем нулевую гипотезу H0 в пользу H1: μ > 10 на уровне значимости 0.05")
else:
    print("Не можем отклонить нулевую гипотезу H0 на уровне значимости 0.05")
```

2. [# 15] Литейный завод производит заливки коленчатых валов из чугуна. Температуры заливки десяти коленчатых валов приведены в таблице. Заданная температура заливки составляет 1400 градусов. Требуется провести тест при уровне α=0.01, чтобы определить, отличается ли температура заливки от заданной.<br>
1395	1394	1396	1438	1404<br>
1403	1406	1401	1400	1401<br>
а. Сформулируйте соответствующую нулевую гипотезу и альтернативную гипотезу.<br>
б. Вычислите значение тестовой статистики и сообщите о своем выводе. Вы можете принять нормальное модельное распределение.

```python
import numpy as np
from scipy.stats import t

# Данные о температуре заливки
temperatures = np.array([1395, 1394, 1396, 1438, 1404, 1403, 1406, 1401, 1400, 1401])

# Параметры
mu0 = 1400  # Заданная температура
alpha = 0.01  # Уровень значимости
n = len(temperatures)  # Размер выборки

# Выборочное среднее и выборочное стандартное отклонение
sample_mean = np.mean(temperatures)
sample_std = np.std(temperatures, ddof=1)

# Вычисление t-статистики
t_stat = (sample_mean - mu0) / (sample_std / np.sqrt(n))

# Количество степеней свободы
df = n - 1

# Критическое значение для двустороннего t-теста
t_critical = t.ppf(1 - alpha/2, df)

# p-значение для двустороннего t-теста
p_value = (1 - t.cdf(abs(t_stat), df)) * 2

# Вывод результатов
print("a.")
print("Нулевая гипотеза H0: средняя температура заливки = 1400 градусов")
print("Альтернативная гипотеза H1: средняя температура заливки ≠ 1400 градусов")

print("\nb.")
print(f"Выборочное среднее: {sample_mean:.2f}")
print(f"Выборочное стандартное отклонение: {sample_std:.2f}")
print(f"t-статистика: {t_stat:.4f}")
print(f"Критическое значение: ±{t_critical:.4f}")
print(f"p-значение: {p_value:.4f}")

if abs(t_stat) > t_critical:
    print("Отклоняем нулевую гипотезу H0 на уровне значимости 0.01")
else:
    print("Не можем отклонить нулевую гипотезу H0 на уровне значимости 0.01")
```

3. [# 15] Рассмотрим данные по теплотворной способности угля <br>

30.880	31.125			<br>
Этот набор данных смоделирован как реализация случайной выборки из распределения N(μ,σ^2), где μ и σ неизвестны. Мы планируем приобрести партию угля, если общая теплотворная способность превышает 31.00 МДж/кг. Проверьте t-критерий для нулевой гипотезы H_0:μ=31.00 против H_1:μ>31.00, используя уровень значимости 0.01, т.е. вычислите значение тестовой статистики, критическое значение теста, и сообщите о своем заключении.

```python
data = np.array([30.990, 31.030, 31.060, 30.921, 30.920,
                 30.990, 31.024, 30.929, 31.050, 30.991,
                 31.208, 30.830, 31.330, 30.810, 31.060,
                 30.800, 31.091, 31.170, 31.026, 31.020,
                 30.880, 31.125])

# Параметры
mu0 = 31.00  # Заданная температура
alpha = 0.01  # Уровень значимости
n = len(data)  # Размер выборки

# Выборочное среднее и выборочное стандартное отклонение
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

# Вычисление t-статистики
t_stat = (sample_mean - mu0) / (sample_std / np.sqrt(n))

# Количество степеней свободы
df = n - 1

# Критическое значение для правостороннего t-теста
t_critical = t.ppf(1 - alpha, df)

# p-значение для правостороннего t-теста
p_value = 1 - t.cdf(t_stat, df)

# Вывод результатов
print("Нулевая гипотеза H0: средняя теплотворная способность = 31.00 МДж/кг")
print("Альтернативная гипотеза H1: средняя теплотворная способность > 31.00 МДж/кг")

print("\nРезультаты теста:")
print(f"Выборочное среднее: {sample_mean:.3f}")
print(f"Выборочное стандартное отклонение: {sample_std:.3f}")
print(f"t-статистика: {t_stat:.4f}")
print(f"Критическое значение (правосторонний тест): {t_critical:.4f}")
print(f"p-значение: {p_value:.4f}")

if t_stat > t_critical:
    print("Отклоняем нулевую гипотезу H0 в пользу H1 на уровне значимости 0.01")
else:
    print("Не можем отклонить нулевую гипотезу H0 на уровне значимости 0.01")
```

[# 20] В ходе исследования, посвященного влиянию утепления стен было измерено еженедельное потребление газа и средняя температура наружного воздуха за 26 недель до и 30 недель после установки утеплителя стен. Термостат в доме был установлен на 20 градусов по Цельсию. Данные приведены в файле temperature_gas.xlsx <br>
Мы моделировали данные до утепления с помощью простой модели линейной регрессии с нормально распределенными ошибками и потреблением газа в качестве переменной реакции<br>
Y_i=α+βx_i+U_i.<br>
Аналогичная модель использовалась для данных после утепления.<br>
a. Используйте данные до изоляции, чтобы выяснить, приводит ли снижение температуры наружного воздуха к увеличению расхода газа. Сформулируйте правильную нулевую и альтернативную гипотезы, вычислите значение тестовой статистики и сообщите о своем выводе, используя уровень значимости 0.05.<br>
б. Проделайте то же самое с данными после изоляции.<br>

```python
import pandas as pd
import numpy as np
from scipy import stats

d = pd.read_excel('temperature_gas.xlsx')

d_b = d.iloc[:26] 
d_a = d.iloc[26:]

def l_r(x, y):
    x_m = np.mean(x)
    y_m = np.mean(y)
    
    b1 = np.sum((x - x_m) * (y - y_m)) / np.sum((x - x_m) ** 2)
    b0 = y_m - b1 * x_m
    
    return b0, b1

def c_s(x, y, b0, b1):
    n = len(x)
    y_p = b0 + b1 * x
    r = y - y_p
    sse = np.sum(r ** 2)
    sst = np.sum((y - np.mean(y)) ** 2)
    r_s = 1 - (sse / sst)
    
    s_e_b1 = np.sqrt(sse / (n - 2)) / np.sqrt(np.sum((x - np.mean(x)) ** 2))
    t_s = b1 / s_e_b1
    p_v = 2 * (1 - stats.t.cdf(np.abs(t_s), df=n-2))
    
    return r_s, s_e_b1, t_s, p_v

X_b = d_b['Temperature']
Y_b = d_b['Consumption']

b0_b, b1_b = l_r(X_b, Y_b)
r_s_b, s_e_b1_b, t_s_b, p_v_b = c_s(X_b, Y_b, b0_b, b1_b)

print(f"P-значение: {p_v_b}")

if p_v_b < 0.05:
    print("Температура влияет на расход газа до утепления.")
else:
    print("Температура не влияет на расход газа до утепления.")

X_a = d_a['Temperature'] 
Y_a = d_a['Consumption']

b0_a, b1_a = l_r(X_a, Y_a)
r_s_a, s_e_b1_a, t_s_a, p_v_a = c_s(X_a, Y_a, b0_a, b1_a)

print(f"\nP-значение: {p_v_a}")

if p_v_a < 0.05:
    print("Температура влияет на расход газа после утепления.") 
else:
    print("Температура не влияет на расход газа после утепления.")
    
```
